```python
#1>> write a function that takes list as input and return all numeric values
def numeric(l):
    l1=[]
    for i in l:
    #if type(i) in [int,float]:
        if type(i)==int or type(i)==float:
             l1.append(i)
    return l1             
```

```python
numeric([1,2,'pwskills',4.5,6.7,'data',9 ,[1,2],(3,4),{5:6},5.6,7000])
#we can pass anything in the list ,it can be int ,float ,string ,list ,tuple ,dictionary etc
#can we write a function which can all the numeric values in list and nested list and other ds passed into the outer list ?
```

```python
#getting all the numeric values even from nested list and tuples etc
def numeric_all(l2):
    l3=[]
    for i in l2:
        if type(i)==list or type(i)==tuple:
                for j in i:
                    if type(j)==int or type(j)==float:
                        l3.append(j)
        elif type(i)==int or type(i)==float:
            l3.append(i)
    return l3
```

```python
numeric_all([1,2,'pwskills',4.5,6.7,'data',9 ,[1,2],(3,4),{5:6},5.6,7000, [10,11],(12,13,14)])
```

```python
#variable length arguement >> when we don't know the number of arguements to be passed in function
#def var_len(*args): #*args is used to take variable length arguements
def summm(*args): #takes any number or arguements and return thier sum
    total=0
    for i in args:
        total+=i
    return total
```

```python
summm(1,2,3,4,5)
summm(10,20)
summm(100,200,300,400,500,600,700,800,10000000000000)
```

```python
#write a function which takes any no. of values and returns the list 
def returnn_list (*args):
    l=[]
    for i in args:
        if type(i)==list or type(i)==tuple:
            for j in i:
                l.append(j)
        l.append(i)
    return l
 
```

```python
returnn_list(1,2,[3,4],(5,6),7,8)
```
