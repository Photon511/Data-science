```python
#generator function - is a special type of function that returns an iterator object
# instead of returning a single value, it uses the yield keyword to produce a series of values
# one at a time, allowing you to iterate over them.
# When a generator function is called, it does not execute the function body immediately.
# Instead, it returns a generator object that can be iterated over to retrieve the values produced by the function.
# Each time the next() function is called on the generator object, the function body is executed
# until it reaches a yield statement, at which point it produces a value and pauses its execution,
# saving its state for the next call to next().
# When next() is called again, the function resumes execution from where it left off, continuing until it reaches the next yield statement or the end of the function.
# This allows for efficient memory usage, as values are generated on-the-fly and do not need to be stored in memory all at once.
```

```python
#regular function -calculates the squares of values and return in one go 
#since its returning all values at once its uses more memory
#if n is large it may lead to memory error and take a lot of time to compute
def square_num(n):
    result = []
    for i in range(n):
        result.append(i ** 2)
    return result
square_num(10)
```

```python
#generator functions are used to avoid this issue
#yield keyword is used  by generator function instead of return statement ,to produce a series of values one at a time
def generate_squares(n):
    for i in range(n):
        yield i ** 2    
generate_squares(10) #it is an iterator object
#calling generator function does not execute the function body immediately
#it returns a generator object that can be iterated over to retrieve the values produced by the     
# generator function
gen = generate_squares(10) #creates generator object and assigns to variable gen
print(next(gen)) #gives first square value 0
print(next(gen)) #gives second square value 1
print(next(gen)) #gives third square value 4
# and so on until all square values are traversed and then it raises StopIteration exception to signal end of iteration      
```

```python
genn = generate_squares(1000000)
for _ in range(10):
    print(next(genn))
#this will not lead to memory error as generator function generates values on-the-fly
# Thus generator functions are more memory efficient than regular functions when dealing with large datasets
```

```python
#fibonacci series using generator function
def fibonacci_gen(n):
    a, b = 0, 1
    for _ in range(n):
        yield a # a beacuse we have to return a first among them at each step
        a, b = b, a + b
f=fibonacci_gen(100000)  
```

```python
next(f)  #0
next(f)  #1
next(f)  #1
next(f)  #2
```
